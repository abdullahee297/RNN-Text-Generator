```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
input_text = "This is the simplest type of neural network architecture where there is a single input and a single output. It is used for straightforward classification tasks such as binary classification where no sequential data is involved."
```

```python
chars = sorted(list(set(input_text)))
char_to_index = {char: i for i, char in enumerate(chars)}
index_to_char = {i: char for i, char in enumerate(chars)}
print(chars)
print(char_to_index)
print(index_to_char)
```

```python
seq_length = 10
sequence = []
labels = []

for i in range(len(input_text) - seq_length):
    seq = input_text[i:i + seq_length]
    label = input_text[i + seq_length]
    sequence.append([char_to_index[char] for char in seq])
    labels.append(char_to_index[label])

x = np.array(sequence)
y = np.array(labels)
# print(y) 
```

```python
x_on_hot = tf.one_hot(x, len(chars))
y_on_hot = tf.one_hot(y, len(chars))
```

```python
model = Sequential()
model.add(SimpleRNN(50, input_shape=(seq_length, len(chars)), activation='relu'))
model.add(Dense(len(chars), activation='softmax'))
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(x_on_hot, y_on_hot, epochs=100)
```

```python
start_seq = "This is the simplest"
generated_text = start_seq

for i in range(100):
    x = np.array([[char_to_index[char] for char in generated_text[-seq_length:]]])
    x_one_hot = tf.one_hot(x, len(chars))
    prediction = model.predict(x_one_hot)
    next_index = np.argmax(prediction)
    next_char = index_to_char[next_index]
    generated_text += next_char

print("Generated Text:")
print(generated_text)
```
